**Setting up**

```python
import pandas as pd
import os
from dotenv import load_dotenv
```

```python
# Connecting sql via pandas using sqlalchemy

from sqlalchemy import create_engine

load_dotenv()
DB_STRING = os.getenv("DB_STRING")

if DB_STRING is None:
    raise ValueError("DB_STRING is not set in the environment.")

db = create_engine(DB_STRING)
```

```python
# Query King County housing sales joined with house details, and load into a DataFrame
query_string = """
select *
FROM eda.king_county_house_sales AS t2
LEFT JOIN eda.king_county_house_details AS t1
ON t2.house_id = t1. id;
"""

df = pd.read_sql(query_string, db)
df.head()
```

```python
# save as CSV
df.to_csv("data/king_county_house.csv", index=False)
```

**Dealing with Missing Values**

1. Waterfront

```python
#counting missing values per column

df.isna().sum()
```

```python
# missing values in waterfront(2391): probably forgot to write 0
# compare the housing price to the price of waterfront housing before replace the missing cells with 0
```

```python
# 1) Check price stats for houses with missing (NaN) waterfront value
df[df["waterfront"].isna()]["price"].agg(["count", "mean", "median"])
```

```python
# 2) Check price stats for houses with waterfront = 1 (has waterfront)
df[df["waterfront"] == 1]["price"].agg(["count", "mean", "median"])
```

```python
# 3) Check price stats for houses with waterfront = 0 (no waterfront)
df[df["waterfront"] == 0]["price"].agg(["count", "mean", "median"])
```

```python
# All three groups (missing data, waterfront 0/1) together (missing values included as their own group)
df.groupby(df["waterfront"], dropna=False)["price"].agg(["count", "mean", "median"])

#-> average values of missling values are close to the ones of 0.
```

```python
#replacing the empty cell with 0
df = df.fillna({"waterfront": 0})
```

```python
#count the missing values in waterfront
df["waterfront"].isna().sum()
```

2. View

```python
#count the empty cells in view : 0
df["view"].isna().sum()
```

```python
# 63 is only 0.29 % of total -> delete them, or replace them with 0
#replacing the empty cell with 0
df = df.fillna({"view": 0})
```

3. yr_renovated

Assumption 1. They forgot to put 0 

Assumption 2. The houses that are not renovated are mostly built recently

```python
# Checking the average missing values of year of the house built
df[df["yr_renovated"].isna()]["yr_built"].agg(["count", "mean", "median"])

# -> mean and median is more less similar, which means the average year_built is somewhere 1970 (apposite to my assumption)
```

```python
#I decided that missing value in yr_renovated should be 0, 
# because there could be other reasons why the history of renovation was not updated. (e.g. partial renovation)
# I think keeping the data as 0 is more reasonble than deleting them

# replacing the empty cell with 0

df = df.fillna({"yr_renovated": 0})
```

```python
df["yr_renovated"].isna().sum()
```

3. Basement

```python
# does missing data in sqft_basement mean those are houses without basement?
# If that is the case, sqft_living = sqft_above
df[df["sqft_basement"].isna()][["sqft_living", "sqft_above"]].head(452)

# -> some of them match, some not
```

```python
# basement = sqft_living - sqft_above

df["sqft_basement"] = df["sqft_basement"].fillna(df["sqft_living"] - df["sqft_above"])
df["sqft_basement"].isna().sum()
```

Rounding data 

```python
# rounding data
df["price"] = df.price.round(0)
df["floors"] = df.floors.round(2)
df["bathrooms"] = df.bathrooms.round(3)


df.head(10)
```

```python
# Convert Zipcode into Integer 
df["zipcode"] = df["zipcode"].astype(int)
df.head()
```

```python
# Save the cleaned dataset 
df.to_csv("data/king_county_house_cleaned.csv", index=False)
```
