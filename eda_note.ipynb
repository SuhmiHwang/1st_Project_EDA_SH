```python
#Connecting sql via pandas using sqlalchemy

from sqlalchemy import create_engine

load_dotenv()
DB_STRING = os.getenv("DB_STRING")

if DB_STRING is None:
    raise ValueError("DB_STRING is not set in the environment.")

db = create_engine(DB_STRING)
```

```python
query_string = """
select *
FROM eda.king_county_house_sales AS t2
LEFT JOIN eda.king_county_house_details AS t1
ON t2.house_id = t1. id;
"""

df = pd.read_sql(query_string, db)
df.head()
```
